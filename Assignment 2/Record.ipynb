```python
import os
from datetime import datetime

file_name = "student_records.txt"

if os.path.exists(file_name):
    print("File already exists.")
else:
    print("New File Created.")


print("\nEnter Student Details")

name = input("Enter Student Name: ")
roll = input("Enter Roll Number: ")
course = input("Enter Course Name: ")

current_time = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

with open(file_name, "w") as file:
    file.write("===== Student Records =====\n")
    file.write(f"Name       : {name}\n")
    file.write(f"Roll No    : {roll}\n")
    file.write(f"Course     : {course}\n")
    file.write(f"Added On   : {current_time}\n")
    file.write("---------------------------------\n")

print("\nRecord saved successfully!")


print("\nStored Records:")
with open(file_name, "r") as file:
    print(file.read())


choice = input("\nDo you want to add another student? (yes/no): ").lower()

while choice == "yes":
    name = input("Enter Student Name: ")
    roll = input("Enter Roll Number: ")
    course = input("Enter Course Name: ")

    current_time = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    with open(file_name, "a") as file:
        file.write(f"Name       : {name}\n")
        file.write(f"Roll No    : {roll}\n")
        file.write(f"Course     : {course}\n")
        file.write(f"Added On   : {current_time}\n")
        file.write("---------------------------------\n")

    print("Student record added successfully!")

    choice = input("\nDo you want to add another student? (yes/no): ").lower()


print("\nFinal Student Records:")
with open(file_name, "r") as file:
    print(file.read())


if os.path.exists(file_name):
    print("File Status: File exists.")
    print("File Size:", os.path.getsize(file_name), "bytes")
else:
    print("File does not exist.")
```
